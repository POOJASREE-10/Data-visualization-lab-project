```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("/content/Food_Waste_Prediction_Dataset.csv")

print(df)

print(df.head())

print(df.shape)

print(df.dtypes)

print(df.describe())

print(df[df["Waste_Amount_kg"] > 10])

print(df[df["Customers"] > 100])

print(df.sort_values("Waste_Amount_kg", ascending=False))

print(df.sort_values("Customers", ascending=False))

print(df.groupby("City")["Waste_Amount_kg"].mean())

print(df.groupby("Food_Item")["Waste_Amount_kg"].mean())

print("Mean Waste:", df["Waste_Amount_kg"].mean())

print("Median Waste:", df["Waste_Amount_kg"].median())

print("Minimum Waste:", df["Waste_Amount_kg"].min())

print("Maximum Waste:", df["Waste_Amount_kg"].max())

print("Standard Deviation:", df["Waste_Amount_kg"].std())

print(df[
    [
        "Prepared_Quantity",
        "Sold_Quantity",
        "Customers",
        "Waste_Amount_kg"
    ]
].corr())

print(
    "Correlation:",
    df["Prepared_Quantity"].corr(df["Waste_Amount_kg"])
)

plt.bar(
    df["Restaurant_Name"].head(20),
    df["Waste_Amount_kg"].head(20)
)

plt.title("Restaurant-wise Food Waste")
plt.xlabel("Restaurant Name")
plt.ylabel("Waste Amount (kg)")
plt.xticks(rotation=90)
plt.show()

city_waste = df.groupby("City")["Waste_Amount_kg"].sum()

plt.plot(
    city_waste.index,
    city_waste.values,
    marker="o"
)

plt.title("City-wise Food Waste")
plt.xlabel("City")
plt.ylabel("Waste Amount (kg)")
plt.xticks(rotation=45)
plt.show()

plt.scatter(
    df["Prepared_Quantity"],
    df["Waste_Amount_kg"]
)

plt.title("Prepared Quantity vs Food Waste")
plt.xlabel("Prepared Quantity")
plt.ylabel("Waste Amount (kg)")
plt.show()

plt.hist(
    df["Waste_Amount_kg"],
    bins=5
)

plt.title("Food Waste Distribution")
plt.xlabel("Waste Amount (kg)")
plt.ylabel("Frequency")
plt.show()

plt.boxplot(df["Waste_Amount_kg"])

plt.title("Food Waste Box Plot")
plt.ylabel("Waste Amount (kg)")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

path = "/content/Food_Waste_Prediction_Dataset.csv"

print("Dataset Path:")

print(path)

print("\nCSV File Found:")

print(os.path.basename(path))

df = pd.read_csv(
    path,
    encoding="latin1",
    low_memory=False
)

print("\nDataset Loaded Successfully")

# ============================================================

# PROGRAM 1 - DATA INSPECTION

# ============================================================

print("\nFirst 5 Records")

print(df.head())

print("\nLast 5 Records")

print(df.tail())

# ============================================================

# PROGRAM 2 - SHAPE

# ============================================================

print("\nShape of Dataset")

print(df.shape)

print("\nRows =", df.shape[0])

print("Columns =", df.shape[1])

# ============================================================

# PROGRAM 3 - DATA TYPES

# ============================================================

print("\nData Types")

print(df.dtypes)

print("\nDataset Info")

df.info()

# ============================================================

# PROGRAM 4 - FILTERING

# Food Waste Greater Than 10 kg

# ============================================================

high_waste = df[df["Waste_Amount_kg"] > 10]

print("\nRecords with Food Waste Greater Than 10 kg")

print(
    high_waste[
        [
            "Restaurant_Name",
            "City",
            "Food_Item",
            "Prepared_Quantity",
            "Sold_Quantity",
            "Waste_Amount_kg"
        ]
    ].head(10)
)

# ============================================================

# PROGRAM 5 - FILTER HIGH CUSTOMER COUNT

# ============================================================

high_customers = df[df["Customers"] > 100]

print("\nRecords with Customers Greater Than 100")

print(
    high_customers[
        [
            "Restaurant_Name",
            "City",
            "Food_Item",
            "Customers",
            "Prepared_Quantity",
            "Waste_Amount_kg"
        ]
    ].head(10)
)
```

```python
# ============================================================

# PROGRAM 6 - SORTING

# Sort by Waste Amount

# ============================================================

sorted_df = df.sort_values(
    by="Waste_Amount_kg",
    ascending=False
)

print("\nTop Records by Food Waste Amount")

print(
    sorted_df[
        [
            "Restaurant_Name",
            "City",
            "Food_Item",
            "Prepared_Quantity",
            "Sold_Quantity",
            "Waste_Amount_kg"
        ]
    ].head(10)
)

# ============================================================

# BAR CHART

# TOP 10 RECORDS BY FOOD WASTE

# ============================================================

top10 = sorted_df.head(10)

plt.figure(figsize=(10, 5))

plt.bar(
    range(len(top10)),
    top10["Waste_Amount_kg"]
)

plt.xticks(
    range(len(top10)),
    top10["Restaurant_Name"],
    rotation=90
)

plt.xlabel("Restaurant")

plt.ylabel("Waste Amount (kg)")

plt.title("Top 10 Restaurants by Food Waste")

plt.show()

# ============================================================

# PROGRAM 7 - STATISTICAL ANALYSIS

# ============================================================

print("\nMean Food Waste =")

print(df["Waste_Amount_kg"].mean())

print("\nMedian Food Waste =")

print(df["Waste_Amount_kg"].median())

print("\nMode Food Waste =")

print(df["Waste_Amount_kg"].mode())

print("\nStandard Deviation =")

print(df["Waste_Amount_kg"].std())

print("\nMinimum Food Waste =")

print(df["Waste_Amount_kg"].min())

print("\nMaximum Food Waste =")

print(df["Waste_Amount_kg"].max())

# ============================================================

# PROGRAM 8 - DESCRIPTIVE STATISTICS

# ============================================================

print("\nDescriptive Statistics")

print(
    df[
        [
            "Prepared_Quantity",
            "Sold_Quantity",
            "Remaining_Quantity",
            "Customers",
            "Discount_Percent",
            "Waste_Amount_kg"
        ]
    ].describe()
)

# ============================================================

# PROGRAM 9 - GROUPING

# Average Food Waste by City

# ============================================================

group = df.groupby(
    "City"
)["Waste_Amount_kg"].mean().sort_values(ascending=False)

print("\nAverage Food Waste by City")

print(group)

# Visualization

group.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.xlabel("City")

plt.ylabel("Average Food Waste (kg)")

plt.title("Average Food Waste by City")

plt.xticks(rotation=45)

plt.show()

# ============================================================

# PROGRAM 10 - AGGREGATION

# ============================================================

summary = df.groupby("City").agg({

    "Waste_Amount_kg": "mean",

    "Prepared_Quantity": "mean",

    "Sold_Quantity": "mean",

    "Customers": "mean"

})

print("\nCity-wise Aggregation")

print(summary)
```

```python
# ============================================================

# PROGRAM 11 - CORRELATION

# ============================================================

numeric = df.select_dtypes(
    include=np.number
)

corr = numeric.corr()

print("\nCorrelation Matrix")

print(corr)

plt.figure(figsize=(12, 8))

sns.heatmap(
    corr,
    annot=False,
    cmap="coolwarm"
)

plt.title("Food Waste Prediction Dataset Correlation Heatmap")

plt.show()

# ============================================================

# PROGRAM 12 - SCATTER PLOT

# Prepared Quantity vs Waste Amount

# ============================================================

plt.figure(figsize=(7, 5))

plt.scatter(
    df["Prepared_Quantity"],
    df["Waste_Amount_kg"]
)

plt.xlabel("Prepared Quantity")

plt.ylabel("Waste Amount (kg)")

plt.title("Prepared Quantity vs Food Waste")

plt.show()

# ============================================================

# PROGRAM 13 - HISTOGRAM

# Food Waste Distribution

# ============================================================

plt.figure(figsize=(7, 5))

plt.hist(
    df["Waste_Amount_kg"].dropna(),
    bins=30
)

plt.xlabel("Waste Amount (kg)")

plt.ylabel("Frequency")

plt.title("Distribution of Food Waste")

plt.show()

# ============================================================

# PROGRAM 14 - BOXPLOT

# ============================================================

plt.figure(figsize=(7, 5))

sns.boxplot(
    x=df["Waste_Amount_kg"].dropna()
)

plt.title("Food Waste Boxplot")

plt.xlabel("Waste Amount (kg)")

plt.show()

# ============================================================

# PROGRAM 15 - VALUE COUNTS

# ============================================================

print("\nRestaurant Counts")

print(
    df["Restaurant_Name"].value_counts()
)

# ============================================================

# PROGRAM 16 - UNIQUE VALUES

# ============================================================

print("\nUnique Restaurants")

print(
    df["Restaurant_Name"].unique()
)

print("\nUnique Cities")

print(
    df["City"].unique()
)

print("\nUnique Cuisine Types")

print(
    df["Cuisine"].unique()
)

print("\nUnique Food Items")

print(
    df["Food_Item"].unique()
)

# ============================================================

# PROGRAM 17 - TOP 10 RESTAURANTS BY FOOD WASTE

# ============================================================

print("\nTop 10 Restaurants by Food Waste")

restaurant_waste = (
    df.groupby("Restaurant_Name")["Waste_Amount_kg"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(restaurant_waste)

# ============================================================

# PROGRAM 18 - FOOD WASTE BY CITY

# ============================================================

print("\nFood Waste by City")

city_waste = (
    df.groupby("City")["Waste_Amount_kg"]
    .sum()
    .sort_values(ascending=False)
)

print(city_waste)

# ============================================================

# PROGRAM 19 - TOP 10 FOOD ITEMS BY FOOD WASTE

# ============================================================

print("\nTop 10 Food Items by Food Waste")

food_waste = (
    df.groupby("Food_Item")["Waste_Amount_kg"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(food_waste)

# ============================================================

# PROGRAM 20 - MOST COMMON CUISINES

# ============================================================

print("\nMost Common Cuisine Types")

cuisines = (
    df["Cuisine"]
    .value_counts()
    .head(10)
)

print(cuisines)
```

```python
# ============================================================

# PROGRAM 21 - EXPORT DATA

# ============================================================

df.to_csv(
    "food_waste_data.csv",
    index=False
)

df.to_excel(
    "food_waste_data.xlsx",
    index=False
)

df.to_json(
    "food_waste_data.json",
    orient="records",
    indent=4
)

print("\nCSV, Excel and JSON files exported successfully.")

# ============================================================

# FINAL REPORT

# ============================================================

print("\n==============================================")

print("EXPERIMENT COMPLETED SUCCESSFULLY")

print("==============================================")

print("Dataset Shape :", df.shape)

print(
    "Missing Values :",
    df.isnull().sum().sum()
)

print(
    "Total Records :",
    len(df)
)

print(
    "Total Prepared Quantity :",
    df["Prepared_Quantity"].sum()
)

print(
    "Total Sold Quantity :",
    df["Sold_Quantity"].sum()
)

print(
    "Total Food Waste :",
    df["Waste_Amount_kg"].sum()
)

print(
    "Average Food Waste :",
    df["Waste_Amount_kg"].mean()
)

print(
    "Number of Restaurants :",
    df["Restaurant_Name"].nunique()
)

print(
    "Number of Cities :",
    df["City"].nunique()
)

print(
    "Number of Cuisines :",
    df["Cuisine"].nunique()
)

print(
    "Number of Food Items :",
    df["Food_Item"].nunique()
)

print("Correlation Matrix Generated")

print("==============================================")
```
