```python
import os

import pandas as pd

path = "/content"

print("Dataset path:", path)

print("\nFiles in dataset:")

print(os.listdir(path))

csv_files = [f for f in os.listdir(path) if f.lower().endswith(".csv")]

if not csv_files:
    raise FileNotFoundError("No CSV file found in the dataset path.")

csv_file = csv_files[0]

print("\nCSV file:", csv_file)

df = pd.read_csv(
    os.path.join(path, csv_file),
    encoding="latin1",
    low_memory=False
)

print("\nFirst 5 rows:")

display(df.head())

print("\nLast 5 rows:")

display(df.tail())

print("\nDataset information:")

df.info()

print("\nShape:")

print(df.shape)

print("\nColumns:")

print(df.columns.tolist())

print("\nData types:")

print(df.dtypes)

print("\nStatistical description:")

display(df.describe(include="all").T)

print("\nMissing values:")

display(df.isnull().sum().sort_values(ascending=False))

print("\nDuplicate rows:")

print(df.duplicated().sum())

print("\nRestaurant names:")

display(df["Restaurant_Name"].value_counts())

print("\nCities:")

display(df["City"].value_counts())

print("\nCuisine types:")

display(df["Cuisine"].value_counts())

print("\nFood items:")

display(df["Food_Item"].value_counts())

print("\nWeather conditions:")

display(df["Weather"].value_counts())

print("\nWeekend status:")

display(df["Weekend"].value_counts())

print("\nHoliday status:")

display(df["Holiday"].value_counts())

print("\nTop restaurants based on food waste:")

display(
    df.groupby("Restaurant_Name")["Waste_Amount_kg"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("\nTop cities based on food waste:")

display(
    df.groupby("City")["Waste_Amount_kg"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTop food items based on food waste:")

display(
    df.groupby("Food_Item")["Waste_Amount_kg"]
    .sum()
    .sort_values(ascending=False)
)

print("\nCorrelation:")

display(df.corr(numeric_only=True))

df["Remaining_Quantity_Calculated"] = (
    df["Prepared_Quantity"] - df["Sold_Quantity"]
)

print("\nCalculated Remaining Quantity:")

display(
    df[
        [
            "Prepared_Quantity",
            "Sold_Quantity",
            "Remaining_Quantity_Calculated"
        ]
    ].head()
)

df["Waste_Percentage"] = (
    df["Waste_Amount_kg"] /
    df["Prepared_Quantity"]
) * 100

print("\nWaste Percentage:")

display(
    df[
        [
            "Restaurant_Name",
            "Prepared_Quantity",
            "Waste_Amount_kg",
            "Waste_Percentage"
        ]
    ].head()
)

df.to_excel(
    "food_waste_data.xlsx",
    index=False
)

df.to_json(
    "food_waste_data.json",
    orient="records",
    indent=4
)

df.to_csv(
    "food_waste_data.csv",
    index=False
)

print("\nCSV, Excel and JSON files created successfully.")
```
