```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

uploaded = files.upload()

csv_file = list(uploaded.keys())[0]

df = pd.read_csv(
    csv_file,
    encoding="latin1",
    low_memory=False
)

print("Dataset Loaded Successfully")

print("\nDataset Shape:")

print(df.shape)

print("\nColumn Names:")

print(df.columns.tolist())

# ============================================================
# PROGRAM 1 - LINE CHART
# ============================================================

city_waste = df.groupby("City")["Waste_Amount_kg"].sum()

plt.figure(figsize=(10, 5))

plt.plot(
    city_waste.index,
    city_waste.values,
    marker="o"
)

plt.title("City-wise Food Waste")

plt.xlabel("City")

plt.ylabel("Waste Amount (kg)")

plt.xticks(rotation=45)

plt.show()

# ============================================================
# PROGRAM 2 - BAR CHART
# ============================================================

restaurant_waste = (
    df.groupby("Restaurant_Name")["Waste_Amount_kg"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 5))

plt.bar(
    restaurant_waste.index,
    restaurant_waste.values
)

plt.title("Top 10 Restaurants by Food Waste")

plt.xlabel("Restaurant Name")

plt.ylabel("Waste Amount (kg)")

plt.xticks(rotation=90)

plt.show()

# ============================================================
# PROGRAM 3 - PIE CHART
# ============================================================

food_waste = (
    df.groupby("Food_Item")["Waste_Amount_kg"]
    .sum()
    .sort_values(ascending=False)
    .head(8)
)

plt.figure(figsize=(8, 8))

plt.pie(
    food_waste.values,
    labels=food_waste.index,
    autopct="%1.1f%%"
)

plt.title("Food Waste Distribution by Food Item")

plt.show()

# ============================================================
# PROGRAM 4 - HISTOGRAM
# ============================================================

plt.figure(figsize=(7, 5))

plt.hist(
    df["Waste_Amount_kg"].dropna(),
    bins=10
)

plt.title("Food Waste Distribution")

plt.xlabel("Waste Amount (kg)")

plt.ylabel("Frequency")

plt.show()

# ============================================================
# PROGRAM 5 - SCATTER PLOT
# ============================================================

plt.figure(figsize=(7, 5))

plt.scatter(
    df["Prepared_Quantity"],
    df["Waste_Amount_kg"]
)

plt.title("Prepared Quantity vs Food Waste")

plt.xlabel("Prepared Quantity")

plt.ylabel("Waste Amount (kg)")

plt.show()

# ============================================================
# PROGRAM 6 - BOX PLOT
# ============================================================

plt.figure(figsize=(7, 5))

plt.boxplot(
    df["Waste_Amount_kg"].dropna()
)

plt.title("Food Waste Box Plot")

plt.ylabel("Waste Amount (kg)")

plt.show()

# ============================================================
# PROGRAM 7 - MULTIPLE LINE CHARTS
# ============================================================

sample = df.head(20)

plt.figure(figsize=(10, 5))

plt.plot(
    sample.index,
    sample["Prepared_Quantity"],
    label="Prepared Quantity"
)

plt.plot(
    sample.index,
    sample["Sold_Quantity"],
    label="Sold Quantity"
)

plt.plot(
    sample.index,
    sample["Waste_Amount_kg"],
    label="Food Waste"
)

plt.title("Prepared, Sold and Wasted Food")

plt.xlabel("Record")

plt.ylabel("Quantity / Waste")

plt.legend()

plt.show()
```

```python
import pandas as pd

import matplotlib.pyplot as plt

# Load Food Waste Prediction dataset

df = pd.read_csv(
    "Food_Waste_Prediction_Dataset.csv",
    encoding="latin1",
    low_memory=False
)

# --------------------------------

# 1. Line Chart – Food Waste by City

# --------------------------------

city_waste = df.groupby("City")["Waste_Amount_kg"].sum()

plt.figure(figsize=(10, 5))

plt.plot(
    city_waste.index,
    city_waste.values,
    marker="o"
)

plt.title("Food Waste by City")

plt.xlabel("City")

plt.ylabel("Total Food Waste (kg)")

plt.xticks(rotation=45)

plt.grid(True)

plt.tight_layout()

plt.show()


# --------------------------------

# 2. Bar Chart – Food Waste by Restaurant

# --------------------------------

restaurant_waste = (
    df.groupby("Restaurant_Name")["Waste_Amount_kg"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))

plt.bar(
    restaurant_waste.index,
    restaurant_waste.values
)

plt.title("Food Waste by Restaurant")

plt.xlabel("Restaurant")

plt.ylabel("Total Food Waste (kg)")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()

plt.show()


# --------------------------------

# 3. Pie Chart – Food Item Waste Distribution

# --------------------------------

food_waste = (
    df.groupby("Food_Item")["Waste_Amount_kg"]
    .sum()
    .sort_values(ascending=False)
)

top_food_items = food_waste.head(6)

plt.figure(figsize=(8, 8))

plt.pie(
    top_food_items.values,
    labels=top_food_items.index,
    autopct="%1.1f%%"
)

plt.title("Top 6 Food Items by Waste")

plt.show()
```
