```python
import pandas as pd
import os
from google.colab import files

uploaded = files.upload()

csv_file = list(uploaded.keys())[0]

print("Uploaded File:")

print(csv_file)

df = pd.read_csv(
    csv_file,
    encoding="latin1",
    low_memory=False
)

print("\nDataset Loaded Successfully")

data = {
    "Restaurant": df["Restaurant_Name"].head(5).tolist(),
    "City": df["City"].head(5).tolist(),
    "Food_Item": df["Food_Item"].head(5).tolist(),
    "Prepared_Quantity": df["Prepared_Quantity"].head(5).tolist(),
    "Sold_Quantity": df["Sold_Quantity"].head(5).tolist(),
    "Waste_Amount_kg": df["Waste_Amount_kg"].head(5).tolist()
}

food_waste_df = pd.DataFrame(data)

print("\nOriginal DataFrame:")

print(food_waste_df)

print("\n1. First 5 rows (head):")

print(food_waste_df.head())

print("\n2. Last 2 rows (tail):")

print(food_waste_df.tail(2))

print("\n3. Shape (rows, columns):")

print(food_waste_df.shape)

print("\n4. Column Names:")

print(food_waste_df.columns)

print("\n5. DataFrame Information:")

food_waste_df.info()

print("\n6. Statistical Summary:")

print(food_waste_df.describe())

print("\n7. Records with Waste Amount greater than 10 kg:")

print(
    df[df["Waste_Amount_kg"] > 10][
        [
            "Restaurant_Name",
            "City",
            "Food_Item",
            "Prepared_Quantity",
            "Sold_Quantity",
            "Waste_Amount_kg"
        ]
    ].head()
)
```
