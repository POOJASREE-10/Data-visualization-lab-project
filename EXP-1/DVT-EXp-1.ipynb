```python
import os
import pandas as pd
import numpy as np

path = "/content/Food_Waste_Prediction_Dataset.csv"

print("Dataset path:", path)

df = pd.read_csv(path)

print("\nFirst 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())

print("\nDataset information:")
df.info()

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nStatistical description:")
display(df.describe())

print("\nMissing values:")
display(df.isnull().sum())

df = df.dropna(how="all")

df = df.drop_duplicates()

print("\nDataset after removing blank rows and duplicates:")
print(df.shape)

df["Remaining_Quantity"] = (
    df["Prepared_Quantity"] - df["Sold_Quantity"]
)

print("\nRemaining Quantity:")
display(
    df[
        [
            "Prepared_Quantity",
            "Sold_Quantity",
            "Remaining_Quantity"
        ]
    ].head()
)

df["Waste_Percentage"] = (
    df["Waste_Amount_kg"] /
    df["Prepared_Quantity"]
) * 100

print("\nWaste Percentage:")
display(
    df[
        [
            "Restaurant_Name",
            "Prepared_Quantity",
            "Waste_Amount_kg",
            "Waste_Percentage"
        ]
    ].head()
)

print("\nRestaurant-wise Food Waste:")
display(
    df.groupby("Restaurant_Name")["Waste_Amount_kg"]
    .sum()
    .sort_values(ascending=False)
)

print("\nCity-wise Food Waste:")
display(
    df.groupby("City")["Waste_Amount_kg"]
    .sum()
    .sort_values(ascending=False)
)

print("\nFood-item-wise Food Waste:")
display(
    df.groupby("Food_Item")["Waste_Amount_kg"]
    .sum()
    .sort_values(ascending=False)
)

df["Date"] = pd.to_datetime(df["Date"])

print("\nFood Waste Trend:")
display(
    df.groupby("Date")["Waste_Amount_kg"]
    .sum()
)

print("\nCorrelation:")
display(
    df.corr(numeric_only=True)
)

df.to_csv(
    "Food_Waste_Cleaned.csv",
    index=False
)

df.to_excel(
    "Food_Waste_Cleaned.xlsx",
    index=False
)

df.to_json(
    "Food_Waste_Cleaned.json",
    orient="records",
    indent=4
)

print("\nCSV, Excel and JSON files created successfully.")
```
